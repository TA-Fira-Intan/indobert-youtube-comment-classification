5 — Fungsi Minimal Cleaning

```python
def minimal_cleaning(text):
    text = str(text)
    # 1. Case folding
    text = text.lower()

    # 2. Hapus URL
    text = re.sub(r"http\S+|www\S+", "", text)

    # 3. Normalisasi whitespace
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

6 — Terapkan Minimal Preprocessing

```python
df["Komentar_preprocessed"] = (
    df["Komentar"]
    .apply(minimal_cleaning))

print("Minimal preprocessing selesai.")
```

14 — Distribusi Panjang Komentar

```python
def kategori_panjang(n):
    if n <= 5:
        return "1–5 kata"
    elif n <= 10:
        return "6–10 kata"
    elif n <= 20:
        return "11–20 kata"
    else:
        return ">20 kata"

df["Kategori_Panjang"] = (
    df["jumlah_kata"]
    .apply(kategori_panjang))

tabel_panjang = pd.crosstab(
    df["Kategori_Panjang"],
    df["Kode Label"])

urutan = [
    "1–5 kata",
    "6–10 kata",
    "11–20 kata",
    ">20 kata"]

tabel_panjang = (
    tabel_panjang
    .reindex(urutan)
    .fillna(0)
    .astype(int))

display(tabel_panjang)
```
