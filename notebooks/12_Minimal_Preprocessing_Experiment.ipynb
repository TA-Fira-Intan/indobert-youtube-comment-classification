```python
# ============================================================
# CELL 1 — LOAD MINIMAL SPLITS
# ============================================================

import os
import pandas as pd

REPO_DIR = "/content/TA"

MINIMAL_SPLIT_DIR = os.path.join(
    REPO_DIR,
    "dataset",
    "split minimal"
)

TRAIN_PATH = os.path.join(
    MINIMAL_SPLIT_DIR,
    "train_split_minimal.csv"
)

VAL_PATH = os.path.join(
    MINIMAL_SPLIT_DIR,
    "validation_split_minimal.csv"
)

TEST_PATH = os.path.join(
    MINIMAL_SPLIT_DIR,
    "test_split_minimal.csv"
)

train_df = pd.read_csv(TRAIN_PATH)
val_df = pd.read_csv(VAL_PATH)
test_df = pd.read_csv(TEST_PATH)

print("=" * 70)
print("MINIMAL PREPROCESSING SPLITS")
print("=" * 70)

print("Train      :", train_df.shape)
print("Validation :", val_df.shape)
print("Test       :", test_df.shape)

print("\nColumns:")
print(train_df.columns.tolist())
```

```python
# ============================================================
# CELL 2 — CHECK LABEL DISTRIBUTION
# ============================================================

LABEL_COL = "Kode Label"
TEXT_COL = "Komentar_preprocessed"

print("=" * 70)
print("LABEL DISTRIBUTION")
print("=" * 70)

print("\nTrain:")
print(
    train_df[LABEL_COL]
    .value_counts()
    .sort_index()
)

print("\nValidation:")
print(
    val_df[LABEL_COL]
    .value_counts()
    .sort_index()
)

print("\nTest:")
print(
    test_df[LABEL_COL]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# CELL 3 — LABEL MAPPING
# ============================================================

label2id = {
    "C0": 0,
    "C1": 1,
    "C2": 2,
    "C3": 3,
    "C4": 4
}

id2label = {
    0: "C0",
    1: "C1",
    2: "C2",
    3: "C3",
    4: "C4"
}

for df in [train_df, val_df, test_df]:
    df["label"] = df[LABEL_COL].map(label2id)

print("Label mapping:")
print(label2id)

print("\nMapped labels:")
print(
    train_df["label"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# CELL 4 — CONVERT TO HUGGING FACE DATASET
# ============================================================

from datasets import Dataset

train_dataset = Dataset.from_pandas(
    train_df[[TEXT_COL, "label"]],
    preserve_index=False
)

val_dataset = Dataset.from_pandas(
    val_df[[TEXT_COL, "label"]],
    preserve_index=False
)

test_dataset = Dataset.from_pandas(
    test_df[[TEXT_COL, "label"]],
    preserve_index=False
)

print("Train:", train_dataset)
print("Val  :", val_dataset)
print("Test :", test_dataset)
```

```python
# ============================================================
# CELL 5 — TOKENIZER
# ============================================================

from transformers import AutoTokenizer

MODEL_NAME = "indobenchmark/indobert-large-p2"
MODEL_LABEL = "IndoBERT Large P2"

MAX_LENGTH = 256

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

print("=" * 70)
print("TOKENIZER")
print("=" * 70)

print("Model      :", MODEL_LABEL)
print("Tokenizer  :", tokenizer.__class__.__name__)
print("MAX_LENGTH :", MAX_LENGTH)
```

```python
# ============================================================
# CELL 6 — TOKENIZATION
# ============================================================

def tokenize_function(examples):
    return tokenizer(
        examples[TEXT_COL],
        truncation=True,
        max_length=MAX_LENGTH
    )

train_tokenized = train_dataset.map(
    tokenize_function,
    batched=True
)

val_tokenized = val_dataset.map(
    tokenize_function,
    batched=True
)

test_tokenized = test_dataset.map(
    tokenize_function,
    batched=True
)

print("Tokenization selesai.")

print("\nTrain columns:")
print(train_tokenized.column_names)
```

```python
# ============================================================
# CELL 7 — LOAD INDOBERT LARGE P2
# ============================================================

import torch
import numpy as np

from transformers import (
    AutoModelForSequenceClassification,
    DataCollatorWithPadding
)

MODEL_NAME = "indobenchmark/indobert-large-p2"
MODEL_LABEL = "IndoBERT Large P2"

MAX_LENGTH = 256

label2id = {
    "C0": 0,
    "C1": 1,
    "C2": 2,
    "C3": 3,
    "C4": 4
}

id2label = {
    0: "C0",
    1: "C1",
    2: "C2",
    3: "C3",
    4: "C4"
}

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=5,
    id2label=id2label,
    label2id=label2id
)

print("=" * 70)
print("MODEL LOADED")
print("=" * 70)

print("Model :", MODEL_LABEL)
print("Labels:", model.num_labels)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
# ============================================================
# CELL 10 — METRICS
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    predictions = np.argmax(
        logits,
        axis=1
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            labels,
            predictions,
            average="macro",
            zero_division=0
        )
    )

    accuracy = accuracy_score(
        labels,
        predictions
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1_macro": f1
    }
```

```python
# ============================================================
# CELL 11 — DATA COLLATOR
# ============================================================

data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

print("Data collator ready.")
```

```python

# CELL 12 — TRAINING ARGUMENTS
# view

from transformers import TrainingArguments

OUTPUT_DIR = "/content/minimal_large_p2_model"

training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    num_train_epochs=10,

    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,

    learning_rate=2e-5,
    weight_decay=0.01,

    # --------------------------------------------------------
    # Evaluation & checkpoint
    # --------------------------------------------------------

    eval_strategy="epoch",
    save_strategy="epoch",

    load_best_model_at_end=True,

    metric_for_best_model="f1_macro",
    greater_is_better=True,

    # --------------------------------------------------------
    # Logging
    # --------------------------------------------------------

    logging_strategy="epoch",

    # --------------------------------------------------------
    # Performance
    # --------------------------------------------------------

    fp16=torch.cuda.is_available(),

    # --------------------------------------------------------
    # Reproducibility
    # --------------------------------------------------------

    seed=42,

    # --------------------------------------------------------
    # Disable external logging
    # --------------------------------------------------------

    report_to="none"
)

print("=" * 70)
print("TRAINING CONFIGURATION")
print("=" * 70)

print("Model          :", MODEL_LABEL)
print("MAX_LENGTH     :", MAX_LENGTH)
print("Learning rate  :", 2e-5)
print("Weight decay   :", 0.01)
print("Class weighted : False")
print("Max epochs     :", 10)
print("Early stopping : 2")
print("FP16           :", torch.cuda.is_available())
```

```python
# ============================================================
# TRAINER
# ============================================================

from transformers import Trainer, EarlyStoppingCallback

trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=train_tokenized,
    eval_dataset=val_tokenized,

    processing_class=tokenizer,
    data_collator=data_collator,

    compute_metrics=compute_metrics,

    callbacks=[
        EarlyStoppingCallback(
            early_stopping_patience=2
        )
    ]
)

print("Trainer ready.")
```

```python
# ============================================================
# TRAINING
# ============================================================

print("=" * 70)
print("START TRAINING")
print("=" * 70)

train_result = trainer.train()

print("\nTraining selesai.")
```

```python
# ============================================================
# TEST — BEST MODEL
# ============================================================

test_results = trainer.evaluate(
    eval_dataset=test_tokenized
)

print("=" * 70)
print("TEST RESULT — BEST MODEL")
print("=" * 70)

for key, value in test_results.items():
    if isinstance(value, float):
        print(f"{key:<20}: {value:.6f}")
    else:
        print(f"{key:<20}: {value}")
```

```python
# ============================================================
# TEST PREDICTIONS
# ============================================================

test_output = trainer.predict(
    test_tokenized
)

predictions = np.argmax(
    test_output.predictions,
    axis=1
)

true_labels = test_output.label_ids

print("=" * 70)
print("TEST PREDICTIONS")
print("=" * 70)

print("Jumlah data test :", len(true_labels))
print("Jumlah prediksi  :", len(predictions))
print("Prediction shape :", test_output.predictions.shape)
```

```python
# ============================================================
# VALIDATION EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)

# Prediction validation
val_output = trainer.predict(
    val_tokenized
)

val_predictions = np.argmax(
    val_output.predictions,
    axis=1
)

val_true = val_output.label_ids

# Metrics
val_precision, val_recall, val_f1, _ = (
    precision_recall_fscore_support(
        val_true,
        val_predictions,
        average="macro",
        zero_division=0
    )
)

val_accuracy = accuracy_score(
    val_true,
    val_predictions
)

# ------------------------------------------------------------
# Metrics DataFrame
# ------------------------------------------------------------

val_metrics_df = pd.DataFrame([{
    "Model": MODEL_LABEL,
    "Dataset": "Validation",
    "Accuracy": val_accuracy,
    "Precision": val_precision,
    "Recall": val_recall,
    "F1 Macro": val_f1
}])

print("=" * 70)
print("VALIDATION METRICS")
print("=" * 70)

display(val_metrics_df)

# view 4
```

```python
# ============================================================
# VALIDATION CLASSIFICATION REPORT
# ============================================================

val_report = classification_report(
    val_true,
    val_predictions,
    labels=[0, 1, 2, 3, 4],
    target_names=[
        "C0", "C1", "C2", "C3", "C4"
    ],
    output_dict=True,
    zero_division=0
)

val_report_df = pd.DataFrame(
    val_report
).transpose()

print("=" * 70)
print("VALIDATION CLASSIFICATION REPORT")
print("=" * 70)

display(val_report_df)
```

```python

# VALIDATION CONFUSION MATRIX
# view 5

val_cm = confusion_matrix(
    val_true,
    val_predictions,
    labels=[0, 1, 2, 3, 4]
)

val_cm_df = pd.DataFrame(
    val_cm,
    index=["C0", "C1", "C2", "C3", "C4"],
    columns=["C0", "C1", "C2", "C3", "C4"]
)

print("=" * 70)
print("VALIDATION CONFUSION MATRIX")
print("=" * 70)

display(val_cm_df)
```

```python
# ============================================================
# TEST EVALUATION
# ============================================================

test_output = trainer.predict(
    test_tokenized
)

test_predictions = np.argmax(
    test_output.predictions,
    axis=1
)

test_true = test_output.label_ids

# Metrics
test_precision, test_recall, test_f1, _ = (
    precision_recall_fscore_support(
        test_true,
        test_predictions,
        average="macro",
        zero_division=0
    )
)

test_accuracy = accuracy_score(
    test_true,
    test_predictions
)

# ------------------------------------------------------------
# Metrics DataFrame
# ------------------------------------------------------------

test_metrics_df = pd.DataFrame([{
    "Model": MODEL_LABEL,
    "Dataset": "Test",
    "Accuracy": test_accuracy,
    "Precision": test_precision,
    "Recall": test_recall,
    "F1 Macro": test_f1
}])

print("=" * 70)
print("TEST METRICS")
print("=" * 70)

display(test_metrics_df)

# VIEW 6
```

```python
# ============================================================
# TEST CLASSIFICATION REPORT
# ============================================================

test_report = classification_report(
    test_true,
    test_predictions,
    labels=[0, 1, 2, 3, 4],
    target_names=[
        "C0", "C1", "C2", "C3", "C4"
    ],
    output_dict=True,
    zero_division=0
)

test_report_df = pd.DataFrame(
    test_report
).transpose()

print("=" * 70)
print("TEST CLASSIFICATION REPORT")
print("=" * 70)

display(test_report_df)
```

```python
# ============================================================
# TEST CONFUSION MATRIX
# ============================================================

test_cm = confusion_matrix(
    test_true,
    test_predictions,
    labels=[0, 1, 2, 3, 4]
)

test_cm_df = pd.DataFrame(
    test_cm,
    index=["C0", "C1", "C2", "C3", "C4"],
    columns=["C0", "C1", "C2", "C3", "C4"]
)

print("=" * 70)
print("TEST CONFUSION MATRIX")
print("=" * 70)

display(test_cm_df)
```

```python
# ============================================================
# SAVE ALL MINIMAL PREPROCESSING RESULTS
# ============================================================

import os
import json

RESULTS_DIR = os.path.join(
    REPO_DIR,
    "results",
    "optimization_minimal-preprocessed"
)

os.makedirs(
    RESULTS_DIR,
    exist_ok=True
)

# ============================================================
# 1. METRICS
# ============================================================

final_minimal_metrics = pd.DataFrame([{
    "model": MODEL_LABEL,
    "preprocessing": "minimal",
    "max_length": MAX_LENGTH,
    "learning_rate": 2e-5,
    "weight_decay": 0.01,
    "class_weighted": False,
    "max_epochs": 10,
    "early_stopping_patience": 2,

    "validation_accuracy": val_accuracy,
    "validation_precision": val_precision,
    "validation_recall": val_recall,
    "validation_f1_macro": val_f1,

    "test_accuracy": test_accuracy,
    "test_precision": test_precision,
    "test_recall": test_recall,
    "test_f1_macro": test_f1,

    "best_checkpoint": trainer.state.best_model_checkpoint,
    "best_validation_f1": trainer.state.best_metric
}])

METRICS_PATH = os.path.join(
    RESULTS_DIR,
    "final_minimal_metrics.csv"
)

final_minimal_metrics.to_csv(
    METRICS_PATH,
    index=False
)


# ============================================================
# 2. CLASSIFICATION REPORT
# ============================================================

test_report_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "final_minimal_classification_report.csv"
    )
)


# ============================================================
# 3. CONFUSION MATRIX
# ============================================================

test_cm_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "final_minimal_confusion_matrix.csv"
    )
)


# ============================================================
# 4. CONFIGURATION
# ============================================================

config = {
    "model": MODEL_LABEL,
    "model_name": MODEL_NAME,
    "preprocessing": "minimal",

    "text_column": TEXT_COL,
    "label_column": LABEL_COL,

    "max_length": MAX_LENGTH,
    "learning_rate": 2e-5,
    "weight_decay": 0.01,

    "class_weighted": False,
    "label_smoothing": False,

    "max_epochs": 10,
    "early_stopping_patience": 2,

    "best_checkpoint": trainer.state.best_model_checkpoint,
    "best_validation_f1": float(trainer.state.best_metric)
}

CONFIG_PATH = os.path.join(
    RESULTS_DIR,
    "final_minimal_config.json"
)

with open(
    CONFIG_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        config,
        f,
        indent=4,
        ensure_ascii=False
    )


# ============================================================
# CHECK
# ============================================================

print("=" * 70)
print("RESULTS SAVED")
print("=" * 70)

print("Folder:")
print(RESULTS_DIR)

print("\nFiles:")

for file in sorted(os.listdir(RESULTS_DIR)):
    path = os.path.join(
        RESULTS_DIR,
        file
    )

    size_kb = os.path.getsize(path) / 1024

    print(
        f"{file:<45} {size_kb:.2f} KB"
    )
```
